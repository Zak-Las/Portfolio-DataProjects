# Step 1: Initial Data Exploration and Cleaning

In this notebook, we will load the NYC Yellow Taxi trip data for January 2025. Our goals are:
1.  Load the dataset.
2.  Perform an initial inspection of the data's structure, columns, and basic statistics.
3.  Identify and handle missing values.
4.  Filter out outliers and irrelevant data for our dispatching problem.
5.  Select the columns necessary for building our simulation environment.

```python
import pandas as pd

# Load the dataset
file_path = 'data/yellow_tripdata_2025-01.parquet'
df = pd.read_parquet(file_path)

# Display the first few rows to get an overview
df.head()
```

```python
# Get a concise summary of the dataframe
df.info()
```

```python
# Generate descriptive statistics
df.describe()
```

#### Key Observations from Initial Exploration
1. __Timestamps are Objects__: tpep_pickup_datetime and tpep_dropoff_datetime are indeed loaded as actual dates. Ergo, we can perform time-based calculations with them in this state.
2. __Passenger Count__: The minimum passenger_count is 0. Trips with no passengers are not relevant to our dispatching problem.
3. __Trip Distance__: The minimum trip_distance is 0. These could be cancelled trips or data errors. A taxi moving 0 distance is not a useful trip to model.
4. __Trip Duration (Implicit)__: The describe() output shows some potentially problematic trip durations. For example, the maximum trip_duration is likely very high, representing outliers. We need to calculate this field to be sure.
5. __Fare Amount__: The minimum fare_amount can be negative or zero, which represents disputes, cancellations, or errors. We should only consider trips with a positive fare.


### Step 2.1: Data Cleaning and Filtering

Based on the initial exploration, I need to clean the data. I will remove outliers and invalid data points. Here's what the cells below do:

1. __Calculate Trip Duration__: I first calculate the `trip_duration` in minutes. This is a crucial feature for filtering.
2. __Apply Filters__: Then, I filter the DataFrame to keep only the trips that are useful for our model:
    - Trip duration between 1 and 60 minutes. This removes both extremely short, likely erroneous trips and very long trips that are outliers.
    - trip_distance greater than 0.
    - passenger_count greater than 0.
    - fare_amount greater than 0.
3. __Post-Cleaning Analysis__: Finally, I run info() and describe() again on the new, filtered DataFrame (df_filtered).

```python
# # Convert to datetime objects - Not neded as they are already in datetime format
# df['tpep_pickup_datetime'] = pd.to_datetime(df['tpep_pickup_datetime'])
# df['tpep_dropoff_datetime'] = pd.to_datetime(df['tpep_dropoff_datetime'])

# Calculate trip duration in minutes
df['trip_duration'] = (df['tpep_dropoff_datetime'] - df['tpep_pickup_datetime']).dt.total_seconds() / 60

# Display the first few rows with the new column
df.head()
```

```python
# Get the original number of rows
original_rows = len(df)

# Apply filters
df_filtered = df[
    (df['trip_duration'] >= 1) & (df['trip_duration'] <= 60) &
    (df['trip_distance'] > 0) &
    (df['passenger_count'] > 0) &
    (df['fare_amount'] > 0)
].copy()

# Get the number of rows after filtering
filtered_rows = len(df_filtered)

print(f"Original number of trips: {original_rows}")
print(f"Number of trips after filtering: {filtered_rows}")
print(f"Percentage of data removed: {((original_rows - filtered_rows) / original_rows) * 100:.2f}%")
```

Let's re-examine the data to see the effect of our cleaning operations. The statistics should now look much more reasonable.

### Step 3: Preparing Location Data

Our goal is to use the official NYC Taxi Zones as our discrete locations. This requires us to:
1. Download the taxi zone lookup table.
2. Filter our trip data to a specific, high-density area. We will choose Manhattan as it has the highest volume of taxi trips.
3. Save the final, cleaned dataset for use in our simulation.

Upon inspecting the columns from `df.info()`, we can see `PULocationID` and `DOLocationID`. These are not arbitrary numbers; they correspond to the official NYC Taxi Zones. The city is already divided into meaningful zones like "JFK Airport," "Midtown Center," etc. Using these official zones is far superior to creating our own grid for several reasons:

- __Real-World Relevance__: These zones are used in official city planning and analysis.
- __Simplicity__: It saves us the complex and often arbitrary task of creating our own geographical grid.
- __Richness__: We can get the borough and zone names for each ID, which is invaluable for analysis.

My task, then, is not to create a discretization, but to utilize the existing one. I need to download the lookup table that maps these IDs to their corresponding Borough and Zone names. This will allow us to understand the data better and, importantly, to filter our simulation to a specific area, like Manhattan, where ride-hailing is most dense.

The cells below are set to handle this. Here is the plan for these new cells:

1. __Download Taxi Zone Lookup Table__: A cell will download the `taxi_zone_lookup.csv` file directly into the `data` directory.
2. __Load and Inspect Lookup Table__: I will load this CSV into a new DataFrame and inspect it.
3. __Filter for Manhattan__: For my simulation, focusing on a dense, contiguous area is a good starting point. I will filter the main dataset to include only trips that start and end within Manhattan. This reduces complexity and focuses the agent on the most active region.
4. __Save Cleaned Data__: Finally, I will save this fully cleaned and filtered DataFrame to a new Parquet file. This is a critical step, as it creates a checkpoint. After this checkpoint, the user won't have to re-run all the cleaning steps every time the user works on the project.

```python
import os
import requests

# Download the taxi zone lookup table
lookup_url = 'https://d37ci6vzurychx.cloudfront.net/misc/taxi_zone_lookup.csv'
lookup_filepath = 'data/taxi_zone_lookup.csv'

if not os.path.exists(lookup_filepath):
    print('Downloading taxi zone lookup table...')
    response = requests.get(lookup_url)
    with open(lookup_filepath, 'wb') as f:
        f.write(response.content)
    print('Download complete.')
else:
    print('Taxi zone lookup table already exists.')

# Load the lookup table
df_zones = pd.read_csv(lookup_filepath)
df_zones.head()
```

Now I will filter my dataset to only include trips that start and end in Manhattan.

```python
# Get the LocationIDs for Manhattan
manhattan_zones = df_zones[df_zones['Borough'] == 'Manhattan']['LocationID'].tolist()

# Filter the trips
original_rows = len(df_filtered)
df_manhattan = df_filtered[
    df_filtered['PULocationID'].isin(manhattan_zones) &
    df_filtered['DOLocationID'].isin(manhattan_zones)
].copy()
filtered_rows = len(df_manhattan)

print(f"Number of trips before Manhattan filter: {original_rows}")
print(f"Number of trips after Manhattan filter: {filtered_rows}")
print(f"Percentage of data remaining: {(filtered_rows / original_rows) * 100:.2f}%")
```

Finally, let's save our processed data to a new file. This will be the input for the next notebook where I will build the simulation environment.

```python
# Save the cleaned and filtered data
output_filepath = 'data/yellow_tripdata_2025-01_manhattan.parquet'
df_manhattan.to_parquet(output_filepath)

print(f"Cleaned data saved to {output_filepath}")
print(f"This concludes our data exploration and cleaning phase.")
```
